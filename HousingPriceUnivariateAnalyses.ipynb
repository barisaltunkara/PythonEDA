```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("data/train.csv")
df
```

```python
def unistats(df):
    output_df = pd.DataFrame(columns=["Count", "Missing", "Unique", "Dtype",
                                     "Mean", "Mode", "Min", "25%", "Median",
                                     "75%", "Max", "Std", "Skew", "Kurtosis"])
    
    for col in df:
        if pd.api.types.is_numeric_dtype(df[col]):
            output_df.loc[col] = [df[col].count(), df[col].isnull().sum(), 
                                  df[col].nunique(), df[col].dtype, df[col].mean(), 
                                  df[col].mode().values[0], df[col].min(), 
                                  df[col].quantile(0.25), df[col].median(), 
                                  df[col].quantile(0.75), df[col].max(), df[col].std(), 
                                  df[col].skew(), df[col].kurt()]
        else:
            output_df.loc[col] = [df[col].count(), df[col].isnull().sum(), 
                                  df[col].nunique(), df[col].dtype, 
                                  "-", "-", "-", "-", "-", "-", "-", "-", "-", "-"]
    
    return output_df

# Test the function
# pd.set_option("display.max_rows", 100)
# pd.set_option("display.max_columns", 100)
unistats(df)
```
